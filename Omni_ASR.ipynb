```python
import io
import numpy as np
import soundfile as sf

def audio_example_to_waveform(audio_ex, target_sr=16000):
    """
    audio_ex is typically ds[i]["audio"] after Audio(decode=False)
    Returns: (waveform_np_float32, sr)
    """
    if isinstance(audio_ex, dict):
        # Case A: local path exists
        if audio_ex.get("path"):
            wav, sr = sf.read(audio_ex["path"])
        # Case B: bytes stored in dataset
        elif audio_ex.get("bytes") is not None:
            data = io.BytesIO(audio_ex["bytes"])
            wav, sr = sf.read(data)
        else:
            raise ValueError(f"Audio dict has neither 'path' nor 'bytes': {audio_ex.keys()}")
    else:
        raise TypeError(f"Unexpected audio example type: {type(audio_ex)}")

    # Convert to mono if stereo
    wav = np.asarray(wav)
    if wav.ndim == 2:
        wav = wav.mean(axis=1)

    wav = wav.astype(np.float32)

    # Resample if needed
    if sr != target_sr:
        import torchaudio
        import torch
        wav_t = torch.from_numpy(wav)
        wav_t = torchaudio.functional.resample(wav_t, sr, target_sr)
        wav = wav_t.numpy()
        sr = target_sr

    return wav, sr
```

# Text Normalization

```python
import re
import unicodedata

ARABIC_BLOCKS = r"\u0600-\u06FF\u0750-\u077F\u08A0-\u08FF"

def normalize_ar(text: str) -> str:
    if not isinstance(text, str):
        return ""

    text = unicodedata.normalize("NFKC", text)
    text = text.replace("\u0640", "")  # tatweel

    # remove combining marks (diacritics)
    text = "".join(ch for ch in text if not unicodedata.combining(ch))
    text = re.sub(r"[\u064B-\u065F\u0670]", "", text)

    # Arabic-Indic numbers
    for i, d in enumerate("٠١٢٣٤٥٦٧٨٩"):
        text = text.replace(d, str(i))

    # Alef variants
    text = re.sub(r"[أإآ]", "ا", text)

    # remove punctuation/symbols (keep Arabic + ascii word chars + whitespace)
    text = re.sub(rf"[^{ARABIC_BLOCKS}\w\s]", " ", text)

    # collapse whitespace
    text = re.sub(r"\s+", " ", text).strip()
    return text
```

```python
import numpy as np

def get_audio_waveform_and_sr(ex, audio_key="audio", default_sr=16000):
    """
    Returns (waveform: np.ndarray float32, sr: int)
    Handles HF Audio dicts, path-only dicts, bytes, and array-like.
    """
    a = ex[audio_key]

    # Case A: already array-like waveform
    if isinstance(a, (list, tuple, np.ndarray)):
        return np.asarray(a, dtype=np.float32), default_sr

    # Case B: dict-like
    if isinstance(a, dict):
        # common HF decoded format
        if "array" in a:
            return np.asarray(a["array"], dtype=np.float32), int(a.get("sampling_rate", default_sr))

        # OmniASR-friendly format
        if "waveform" in a:
            wf = a["waveform"]
            wf = wf.detach().cpu().numpy() if hasattr(wf, "detach") else np.asarray(wf)
            return wf.astype(np.float32), int(a.get("sample_rate", a.get("sampling_rate", default_sr)))

        # some datasets use "values"
        if "values" in a:
            return np.asarray(a["values"], dtype=np.float32), int(a.get("sampling_rate", default_sr))

        # load from path if present
        if "path" in a and a["path"]:
            import soundfile as sf
            wf, sr = sf.read(a["path"], dtype="float32")
            if wf.ndim > 1:  # convert to mono
                wf = wf.mean(axis=1)
            return wf, int(sr)

        # load from bytes if present
        if "bytes" in a and a["bytes"]:
            import io, soundfile as sf
            wf, sr = sf.read(io.BytesIO(a["bytes"]), dtype="float32")
            if wf.ndim > 1:
                wf = wf.mean(axis=1)
            return wf, int(sr)

    raise KeyError(f"Don't know how to read `{audio_key}`. Got type={type(a)} keys={list(a.keys()) if isinstance(a, dict) else None}")
```

```python
def ensure_16k(wf, sr, target_sr=16000):
    if sr == target_sr:
        return wf, sr
    import librosa
    wf16 = librosa.resample(wf, orig_sr=sr, target_sr=target_sr)
    return wf16.astype(np.float32), target_sr
```

```python
import time
def evaluate_model(ds_test, transcribe_fn, N=50):
    refs, hyps = [], []
    t0 = time.time()

    for i in range(min(N, len(ds_test))):
        ex = ds_test[i]

        wf, sr = get_audio_waveform_and_sr(ex, audio_key="audio", default_sr=16000)
        wf, sr = ensure_16k(wf, sr, target_sr=16000)

        ref = ex["cleaned_text"] if "cleaned_text" in ex else ex["text"]
        hyp = transcribe_fn(wf, sample_rate=sr)

        refs.append(normalize_ar(ref))
        hyps.append(normalize_ar(hyp))

    dt = time.time() - t0
    return {
        "N": len(refs),
        "WER": wer.compute(predictions=hyps, references=refs),
        "CER": cer.compute(predictions=hyps, references=refs),
        "seconds": dt,
        "sec_per_utt": dt / max(1, len(refs))
    }
```

# Speech-to-Text (LLMs)

## omniASR-LLM (Meta pipeline)


```python
def make_omniasr_transcriber(model_card="omniASR_LLM_7B", lang_code="arb_Arab", batch_size=1):
    import torch
    from omnilingual_asr.models.inference.pipeline import ASRInferencePipeline

    device = "cuda" if torch.cuda.is_available() else "cpu"
    dtype = torch.float16 if device == "cuda" else torch.float32

    pipeline = ASRInferencePipeline(model_card=model_card, device=device, dtype=dtype)

    def _to_item(audio, sample_rate=16000):
        # OmniASR-predecoded dict
        if isinstance(audio, dict) and "waveform" in audio and ("sample_rate" in audio or "sampling_rate" in audio):
            wf = audio["waveform"]
            sr = int(audio.get("sample_rate", audio.get("sampling_rate")))
            if not torch.is_tensor(wf):
                wf = torch.tensor(wf)
            return {"waveform": wf, "sample_rate": sr}

        # path-like: let pipeline decode
        if isinstance(audio, (str, bytes)):
            return audio

        # array-like waveform
        wf = audio
        if not torch.is_tensor(wf):
            wf = torch.tensor(wf)

        if wf.ndim == 2 and wf.shape[0] == 1:
            wf = wf.squeeze(0)
        elif wf.ndim != 1:
            wf = wf.reshape(-1)

        wf = wf.to(dtype=torch.float32, device="cpu")
        return {"waveform": wf, "sample_rate": int(sample_rate)}

    def transcribe(audio, lang=None, sample_rate=16000):
        lc = lang if lang is not None else lang_code
        item = _to_item(audio, sample_rate=sample_rate)
        outs = pipeline.transcribe([item], lang=[lc], batch_size=batch_size)
        return outs[0] if outs else ""

    return transcribe
```

# Runner

```python
MODELS = {
    # omniASR-LLM (Meta pipeline)
    "facebook/omniASR-LLM-300M": "omni_llm",
    "facebook/omniASR-LLM-1B": "omni_llm",
    "facebook/omniASR-LLM-3B": "omni_llm",
    "facebook/omniASR-LLM-7B": "omni_llm",
}
```

```python
def make_transcriber(model_id: str):
    fam = MODELS[model_id]

    if fam == "omni_llm":
        # map HF repo id -> omnilingual-asr model_card name
        hf_to_card = {
            "facebook/omniASR-LLM-300M": "omniASR_LLM_300M",
            "facebook/omniASR-LLM-1B":   "omniASR_LLM_1B",
            "facebook/omniASR-LLM-3B":   "omniASR_LLM_3B",
            "facebook/omniASR-LLM-7B":   "omniASR_LLM_7B",
        }
        return make_omniasr_transcriber(hf_to_card[model_id], lang_code="arb_Arab")

    raise ValueError(f"Unknown family: {fam}")
```

```python
import pandas as pd

def eval_and_show(ds_test, model_id, transcribe_fn, N_eval=20, N_show=3):
    # run metrics (already robust if you updated evaluate_model as shown)
    res = evaluate_model(ds_test, transcribe_fn, N=N_eval)
    res["model"] = model_id

    print("\n" + "="*80)
    print(model_id)
    print(res)

    # show examples (robust audio loading + resample)
    for i in range(min(N_show, len(ds_test))):
        ex = ds_test[i]
        ref_raw = ex["cleaned_text"] if "cleaned_text" in ex else ex["text"]

        wf, sr = get_audio_waveform_and_sr(ex, audio_key="audio", default_sr=16000)
        wf, sr = ensure_16k(wf, sr, target_sr=16000)

        hyp_raw = transcribe_fn(wf, sample_rate=sr)

        print(f"\n[{i}] REF : {ref_raw}")
        print(f"[{i}] HYP : {hyp_raw}")
        print(f"[{i}] NREF: {normalize_ar(ref_raw)}")
        print(f"[{i}] NHYP: {normalize_ar(hyp_raw)}")

    return res
```

```python
quick_set = [
    "facebook/omniASR-LLM-300M",              # omni llm family
]

results = []
for mid in quick_set:
    tr = make_transcriber(mid)
    results.append(eval_and_show(ds["test"], mid, tr, N_eval=10, N_show=2))

df = pd.DataFrame(results).sort_values("WER")
df
```
